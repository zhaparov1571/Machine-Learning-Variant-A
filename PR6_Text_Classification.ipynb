Мини-корпус русскоязычных отзывов с метками: 1 — положительный, 0 — отрицательный. Для реальной работы лучше использовать более крупный датасет.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, f1_score, classification_report

records = [('Мне очень понравилось, отличный товар', 1), ('Качество прекрасное, я доволен', 1), ('Быстрая доставка и хороший сервис', 1), ('Очень хороший результат, рекомендую', 1), ('Все отлично, спасибо большое', 1), ('Мне нравится этот продукт', 1), ('Классная вещь, работает идеально', 1), ('Покупкой доволен, всё супер', 1), ('Хорошее качество за свои деньги', 1), ('Прекрасно работает, всё понравилось', 1), ('Ужасное качество, совсем не понравилось', 0), ('Очень плохой товар', 0), ('Доставка медленная и сервис ужасный', 0), ('Я недоволен покупкой', 0), ('Не рекомендую, всё плохо', 0), ('Продукт сломался сразу', 0), ('Деньги потрачены зря', 0), ('Качество отвратительное', 0), ('Работает плохо, разочарован', 0), ('Очень неприятный опыт', 0), ('Отличный сервис, всё быстро', 1), ('Товар хороший и полезный', 1), ('Очень доволен качеством', 1), ('Всё работает хорошо', 1), ('Покупка превзошла ожидания', 1), ('Приятно удивлён, рекомендую', 1), ('Плохой сервис и плохое качество', 0), ('Совершенно бесполезная вещь', 0), ('Не работает, ужас', 0), ('Покупка разочаровала', 0), ('Всё пришло сломанным', 0), ('Очень долго и неудобно', 0)]
df = pd.DataFrame(records, columns=["text", "label"])
display(df.head())
print("Размер корпуса:", df.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.25, random_state=42, stratify=df["label"]
)

results = []
models = {}
for name, vectorizer in [
    ("CountVectorizer", CountVectorizer()),
    ("TfidfVectorizer", TfidfVectorizer())
]:
    pipe = Pipeline([
        ("vectorizer", vectorizer),
        ("classifier", MultinomialNB())
    ])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    results.append({
        "Векторизация": name,
        "Accuracy": accuracy_score(y_test, pred),
        "F1": f1_score(y_test, pred, zero_division=0),
        "Размерность": pipe.named_steps["vectorizer"].transform(X_train).shape[1]
    })
    models[name] = pipe

results_df = pd.DataFrame(results)
display(results_df)
```

## Вывод
Сравни значения Accuracy и F1 в таблице. На этом небольшом учебном корпусе результат может заметно зависеть от разбиения; нельзя делать общий вывод, что один способ всегда лучше другого.
